# Determine line spacing
Here we estimate the lines spacing for an entire survey, as well as separately calculating the line spacing and tie spacing.

```python
# %load_ext autoreload
# %autoreload 2

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import airbornegeo
```

```python
data_df = pd.read_csv("data/AGAP_magnetic_survey_processed.csv")
data_df = data_df[["easting", "northing", "line", "unixtime"]]

# define flight lines vs tie lines (0=flight, 1=tie)
data_df["line_type"] = np.where(data_df.line >= 142, 1, 0)

data_df
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10, 6))

df = data_df[::10]
ax = df[df.line_type == 0].plot.scatter(
    "easting",
    "northing",
    c="line",
    s=0.02,
    cmap="rainbow",
    ax=axs[0],
    colorbar=False,
    title="Lines",
)
ax.set_xlim(data_df.easting.min(), data_df.easting.max())
ax.set_ylim(data_df.northing.min(), data_df.northing.max())
ax.set_aspect("equal")
plt.colorbar(ax.collections[0], ax=ax, shrink=0.5)
airbornegeo.add_scalebar(ax, 100e3)

ax = df[df.line_type == 1].plot.scatter(
    "easting",
    "northing",
    c="line",
    s=0.02,
    cmap="rainbow",
    ax=axs[1],
    colorbar=False,
    title="Ties",
)
ax.set_xlim(data_df.easting.min(), data_df.easting.max())
ax.set_ylim(data_df.northing.min(), data_df.northing.max())
ax.set_aspect("equal")
plt.colorbar(ax.collections[0], ax=ax, shrink=0.5)
airbornegeo.add_scalebar(ax, 100e3)

plt.tight_layout()
plt.show()
```

```python
# to speed up the calculation, we can decimated the dataset
df = data_df[::10]
```

```python
survey = airbornegeo.Survey(
    df,
    line_column="line",
    line_type_column="line_type",
)

# Access the median_line_spacings cached property
# This automatically computes spacing per line type
spacings = survey.median_line_spacings
print(f"Median line spacings: {spacings}\n")

for line_type_name, spacing in spacings.items():
    print(f"{line_type_name.capitalize()} line spacing: {spacing} m")
```
