# Process the AGAP airborne gravity data

We try to recreate the BAS processing steps, as outlined here: https://data.bas.ac.uk/full-record.php?id=GB/NERC/BAS/PDC/01310

```python
# %load_ext autoreload
# %autoreload 2

import boule as bl
import cmocean
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import verde as vd

import airbornegeo
```

## Load data

This is a subset of the BAS AGAP survey over Antarctica's Gamburtsev Subglacial Mountains. The file is download and subset in the notebook `AGAP_survey`.

```python
data_df = pd.read_csv("data/AGAP_gravity_survey.csv")

print(data_df.columns)
data_df.head()
```

```python
survey = airbornegeo.Survey(
    data_df,
    line_column="line",
    time_column="unixtime",
    height_column="Height_WGS1984",
    latitude_column="Lat",
    longitude_column="Lon",
    distance_column="distance_along_line",
)
survey
```

## Distance along line

```python
survey.along_track_distance()
survey.data[["easting", "northing", "distance_along_line"]].head()
```

## Relative gravity
$$
g_{rel} = s (t +(b k)+c)
$$

where $g_{rel}$ is the relative gravity, $t$ is the true spring tension, $b$ is the beam velocity, $k$ is 30, $c$ is the cross-coupling, and $s$ is the meter scale value, which is 0.9966.

The beam velocity ($b$) is from a centered finite difference  approximation of the raw beam position ($RB$).

```python
survey.data["grav_relative"] = 0.9966 * (
    survey.data["ST_real"] + (survey.data["Beam_vel"] * 30) + survey.data["CC"]
)
survey.data[
    ["unixtime", "RB", "Beam_vel", "CC", "ST", "ST_real", "grav_relative"]
].head()
```

```python
max_abs = vd.maxabs(survey.data.grav_relative, percentile=95)

survey.data[::50].plot.scatter(
    "easting",
    "northing",
    c="grav_relative",
    s=0.6,
    cmap=cmocean.cm.balance,
    vmin=-max_abs,
    vmax=max_abs,
).set_aspect("equal")
```

```python
# for now just use there values
survey.data = data_df.drop(columns=["grav_relative"], errors="ignore")
survey.data["grav_relative"] = data_df.rec_grav
```

## Absolute gravity

Still readings are in mGal and were calculating assuming a 2nd order best fit to the approximately linear drift of the meter observed at the tie down points. For each line, there are 2-5 unique still values. For the whole survey, the base station readings are all equal to 981970.73 mGal.

Absolute gravity is then calculated as:
$$
g_{abs} = g_{rel} - g_{still} + g_{base}
$$

```python
survey.data["grav_absolute"] = (
    survey.data["grav_relative"] - survey.data["Still"] + survey.data["Base"]
)
survey.data[["grav_relative", "Base", "Still", "grav_absolute"]].head()
```

```python
cpt_lims = vd.minmax(survey.data.grav_absolute, min_percentile=5, max_percentile=95)

survey.data[::50].plot.scatter(
    "easting",
    "northing",
    c="grav_absolute",
    s=0.6,
    vmin=cpt_lims[0],
    vmax=cpt_lims[1],
).set_aspect("equal")
```

```python
# for now just use their values
survey.data = data_df.drop(columns=["grav_absolute"], errors="ignore")
survey.data["grav_absolute"] = data_df.Abs_grav
```

## Vertical acceleration corrections

The vertical acceleration is calculated as the 2nd derivative of flight altitude w.r.t. time with a 3 point mean filter applied after differencing to reduce short wavelength noise.

```python
survey.vertical_acceleration(
    smoothing_window=3,
    result_column="vertical_acceleration",
)
# Convert from m/s2 to mGal (already done by the method via result multiplier)
survey.data.describe()
```

```python
df = survey.data[survey.data.line == 1]

ylim = vd.minmax(df.VaccCor)  # , min_percentile=5, max_percentile=95)

ax = df.plot.line(
    "distance_along_line",
    "vertical_acceleration",
    style="bp",
    ms=1,
    ylim=ylim,
)

ax = df.plot.line(
    "distance_along_line",
    "VaccCor",
    style="rx",
    ms=1,
    # ax=ax,
    ylim=ylim,
)
```

```python
survey.data["dif"] = survey.data.VaccCor - survey.data.vertical_acceleration
sns.histplot(survey.data.dif, kde=True)
plt.title(
    f"Vertical acceleration difference with published values, RMSE:{round(airbornegeo.rmse(survey.data.dif), 2)}"
)
plt.show()
```

```python
# for now just use the BAS values
survey.data = data_df.drop(columns=["vertical_acceleration"], errors="ignore")
survey.data["vertical_acceleration"] = data_df.VaccCor
```

## Observed gravity

```python
survey.data["grav_observed"] = (
    survey.data["grav_absolute"]
    - survey.data["vertical_acceleration"]
    + survey.data["eotvos_correction"]
    - (0.5 * survey.data["horizontal_acceleration"])
)
```

```python
cpt_lims = vd.minmax(survey.data.grav_observed, min_percentile=5, max_percentile=95)
survey.data[::50].plot.scatter(
    "easting",
    "northing",
    c="grav_observed",
    s=0.6,
    vmin=cpt_lims[0],
    vmax=cpt_lims[1],
).set_aspect("equal")
```

## Normal gravity corrections 

BAS used a separate latitude correction and free-air correction. We will instead combine them with a Normal gravity correction.

```python
# combine the published Latitude and Free air corrections into a Normal gravity column
survey.data["NormalGrav"] = survey.data.LatCor - survey.data.FaCor
```

```python
# Calculate normal gravity using the WGS84 ellipsoid
ellipsoid = bl.WGS84

survey.data["grav_normal"] = ellipsoid.normal_gravity(
    (None, survey.data.Lat, survey.data.Height_WGS1984)
)
```

```python
survey.data[::50].plot.scatter(
    "easting",
    "northing",
    c="grav_normal",
    s=0.6,
).set_aspect("equal")
```

```python
df = survey.data[survey.data.line == 1]
ax = df.plot.line(
    "distance_along_line",
    "grav_normal",
    style="bp",
    ms=1,
)
ax = df.plot.line(
    "distance_along_line",
    "NormalGrav",
    style="rx",
    ms=1,
    ax=ax,
)
```

```python
survey.data["dif"] = survey.data.NormalGrav - survey.data.grav_normal
sns.histplot(survey.data.dif, kde=True)
plt.title(
    f"Normal gravity difference with published values, RMSE:{round(airbornegeo.rmse(survey.data.dif), 2)}"
)
plt.show()
```

```python
# for now just use the BAS values
survey.data = data_df.drop(columns=["grav_normal"], errors="ignore")
survey.data["grav_normal"] = data_df.NormalGrav
```

## Gravity Distrubance

```python
# The disturbance is the observed minus normal gravity (calculated at the
# observation point)
survey.data["grav_disturbance"] = data_df.grav_observed - data_df.grav_normal
```

```python
cpt_lims = vd.maxabs(survey.data.Free_air, percentile=95)
survey.data[::50].plot.scatter(
    "easting",
    "northing",
    c="grav_disturbance",
    s=0.6,
    cmap=cmocean.cm.balance,
    vmin=-cpt_lims,
    vmax=cpt_lims,
).set_aspect("equal")
```

```python
survey.data["dif"] = survey.data.Free_air - survey.data.grav_disturbance
sns.histplot(survey.data.dif, kde=True)
plt.title(
    f"Gravity disturbance (FAA) difference with published values, RMSE:{round(airbornegeo.rmse(survey.data.dif), 2)}"
)
plt.show()
```

```python
df = survey.data[survey.data.line == 1]

ylim = vd.minmax(df.grav_disturbance)  # , min_percentile=5, max_percentile=95)

ax = df.plot.line(
    "distance_along_line",
    "grav_disturbance",
    style="bp",
    ms=1,
    ylim=ylim,
)

ax = df.plot.line(
    "distance_along_line",
    "Free_air",
    style="rx",
    ms=1,
    ax=ax,
    ylim=ylim,
)
```

## Filtering

Filtered free air anomaly (FAA_filt) used 9 km 1/2 wavelength space domain kernel filter (Holt et al., 2006).

```python
survey.filter_line(
    filter_width=26_000,
    filter_type="lowpass",
    data_column="grav_disturbance",
    result_column="grav_disturbance_filt",
    pad_width_percentage=20,
)
```

```python
cpt_lims = vd.maxabs(survey.data.FAA_filt, percentile=95)
survey.data[::50].plot.scatter(
    "easting",
    "northing",
    c="grav_disturbance_filt",
    s=0.6,
    cmap=cmocean.cm.balance,
    vmin=-cpt_lims,
    vmax=cpt_lims,
).set_aspect("equal")
```

```python
survey.data[::50].plot.scatter(
    "easting",
    "northing",
    c="FAA_filt",
    s=0.6,
    cmap=cmocean.cm.balance,
    vmin=-cpt_lims,
    vmax=cpt_lims,
).set_aspect("equal")
```

```python
survey.data["dif"] = survey.data.FAA_filt - survey.data.grav_disturbance_filt
sns.histplot(survey.data.dif, kde=True)
plt.title(
    f"filtered FAA difference with published values, RMSE:{round(airbornegeo.rmse(survey.data.dif), 2)}"
)
plt.show()
```

```python
cpt_lims = vd.maxabs(survey.data.dif, percentile=95)
survey.data[::50].plot.scatter(
    "easting",
    "northing",
    c="dif",
    s=0.6,
    cmap=cmocean.cm.balance,
    vmin=-cpt_lims,
    vmax=cpt_lims,
).set_aspect("equal")
```
